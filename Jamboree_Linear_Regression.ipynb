Number Of Outliers

```python
# Columns for outlier analysis
numeric_cols = [
    'GRE Score',
    'TOEFL Score',
    'University Rating',
    'SOP',
    'LOR ',
    'CGPA',
    'Chance of Admit '
]
outlier_summary = []

for col in numeric_cols:
  Q1 = df[col].quantile(0.25)
  Q3 = df[col].quantile(0.75)

  IQR = Q3 - Q1

  lower_bound = Q1 - 1.5 * IQR
  upper_bound = Q3 + 1.5 * IQR

  outliers = df[(df[col] < lower_bound) | (df[col] > upper_bound)]

  outlier_summary.append({
      'Column': col,
      'Q1': Q1,
      'Q3': Q3,
      'IQR': IQR,
      'Lower_Bound': lower_bound,
      'Upper_Bound': upper_bound,
      'Outlier Count': len(outliers),
      'Outlier %': round(len(outliers) / len(df) * 100, 2)
  })

outlier_df = pd.DataFrame(outlier_summary)
outlier_df
```

```python
# Separate predictors (x) and target (y)
x = df.drop('Chance of Admit ', axis = 1)
y = df['Chance of Admit ']
```

```python
print('x.shape:', x.shape)
print('y.shape:', y.shape)
```

Train Test Split

```python
from sklearn.model_selection import train_test_split
x_train, x_test, y_train, y_test = train_test_split(
    x,
    y,
    test_size=0.20,
    random_state=42
)
```

```python
# Checking resulting shapes
print('x_train:' , x_train.shape)
print('x_test:' , x_test.shape)
print('y_train:' , y_train.shape)
print('y_test:' , y_test.shape)
```

Model Building

```python
import statsmodels.api as sm

# add constant/intercept to the predictor
x_train_sm = sm.add_constant(x_train)

# build the linear regression model
model = sm.OLS(y_train, x_train_sm).fit()

# display model summary
print(model.summary())
```

```python
import statsmodels.api as sm
import matplotlib.pyplot as plt
import seaborn as sns

# Add constant to the test predictors (matching what was done for train data)
x_test_sm = sm.add_constant(x_test)

# Predictions on test data
y_pred = model.predict(x_test_sm)

plt.figure(figsize=(8,5))

sns.scatterplot(x=y_test, y=y_pred)

plt.xlabel('Actual Chance of Admit')
plt.ylabel('Predicted Chance of Admit')
plt.title('Actual vs Predicted Chance of Admit')

# Perfect prediction line
plt.plot([y_test.min(), y_test.max()],
         [y_test.min(), y_test.max()],
         'r--')

plt.show()
```

```python
from statsmodels.stats.outliers_influence import variance_inflation_factor
import pandas as pd

vif_data = pd.DataFrame()
vif_data['Feature'] = x_train.columns

vif_data['VIF'] = [
    variance_inflation_factor(x_train.values, i)
    for i in range(x_train.shape[1])
]
vif_data
```

Observation: VIF Analysis shows that all predictor variables have VIF values below 5. GRE Score (4.49) and CGPA (4.65) have relatively higher VIF'S, indicating some multicollinearity, but they remain within acceptable threshold. Therefore, no variables were removed based on VIF

Mean Residual Error

```python
residuals = model.resid
print('Mean of Residuals:', residuals.mean())
```

Observation: The mean residual is approximately zero, specifically 3.92 x 10e-16. Therefore, the mean residual assumption is satisfied. This indicates that the model's residuals are centered around zero.

Checking Linearity

```python
fitted_values = model.fittedvalues
residuals = model.resid
```

```python
import matplotlib.pyplot as plt
plt.figure(figsize=(8,5))

plt.scatter(fitted_values, residuals)

plt.axhline(y=0, color='r', linestyle='--')

plt.xlabel('Fitted Values')
plt.ylabel('Residuals')
plt.title('Residuals vs Fitted Values')

plt.show()
```

Observation: The residuals are scattered around zero without a clear curved or systematic pattern, indicating that the linearity assumption was reasonably satisfied.

There is some change in the spread of residuals across fitted values, which may indicate heteroscedasticity. I would investigate that separately using the homoscedasticity checks.

Breusch-Pagan Test

```python
from statsmodels.stats.diagnostic import het_breuschpagan

bp_test = het_breuschpagan(
    model.resid,
    model.model.exog
)

labels = [
    "LM Statistic",
    "LM p-value",
    "F Statistic",
    "F p-value"
]

for label, value in zip(labels, bp_test):
    print(label, ":", value)
```

Observation: The residuals versus fitted values plot showed some variation in residual spread across fitted values. The Breusch-Pagan test produced a p-value below 0.05, indicating statistically significant heteroscedasticity. Therefore, the homoscedasticity assumption is not satisfied.

Checking Normality

Histogram of Residuals

```python
import matplotlib.pyplot as plt

plt.figure(figsize=(8, 5))

plt.hist(residuals, bins=30)

plt.xlabel("Residuals")
plt.ylabel("Frequency")
plt.title("Distribution of Residuals")

plt.show()
```

Observation: Approximately bell-shaped, but with some deviation from perfect normality and a few extreme residuals.

Q-Q Plot

```python
import statsmodels.api as sm
import matplotlib.pyplot as plt

sm.qqplot(residuals, line='45', fit=True)

plt.title("Q-Q Plot of Residuals")
plt.show()
```

Observation: The Q-Q plot shows that most residuals in the middle are reasonably close to the diagonal line, but there are noticeable deviations at both tails, especially the lower/left tail. This indicates that the residuals are not perfectly normally distributed. Therefore, the normality assumption is not fully satisfied

Jarque-Bera Test

```python
from scipy.stats import jarque_bera

jb_stat, jb_pvalue = jarque_bera(residuals)

print("Jarque-Bera Statistic:", jb_stat)
print("Jarque-Bera p-value:", jb_pvalue)
```

Observation: The linearity and mean-residual assumptions are reasonably satisfied, and no severe multicollinearity was observed as all VIF values were below 5. However, the Breusch-Pagan test indicates heteroscedasticity, while the Jarque-Bera test and Q-Q plot indicate that the residuals are not normally distributed.

Model Evaluation

Training Predictions

```python
y_train_pred = model.predict(x_train_sm)
model = sm.OLS(y_train, x_train_sm).fit()
```

Testing Predictions

```python
x_test_sm = sm.add_constant(x_test)
y_test_pred = model.predict(x_test_sm)
```

Calculating MAE, RMSE & R2

```python
from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score
import numpy as np

# Training metrics
train_mae = mean_absolute_error(y_train, y_train_pred)
train_rmse = np.sqrt(mean_squared_error(y_train, y_train_pred))
train_r2 = r2_score(y_train, y_train_pred)

# Testing metrics
test_mae = mean_absolute_error(y_test, y_test_pred)
test_rmse = np.sqrt(mean_squared_error(y_test, y_test_pred))
test_r2 = r2_score(y_test, y_test_pred)

print("Training MAE:", train_mae)
print("Training RMSE:", train_rmse)
print("Training R²:", train_r2)

print("\nTesting MAE:", test_mae)
print("Testing RMSE:", test_rmse)
print("Testing R²:", test_r2)
```

Observation: The model achieved a training MAE of 0.0425 and test MAE of 0.0427, indicating an average prediction error of approximately 0.043 in Chance of Admit. The training and test RMSE values were 0.0594 and 0.0609, respectively. The model achieved an R² of 0.821 on the training set and 0.819 on the test set, indicating that approximately 81.9% of the variation in Chance of Admit is explained by the predictors on unseen data. The close train-test performance suggests that there is little evidence of overfitting

Adjusted R2

Observation: The model achieved an R² of 0.8211 and an Adjusted R² of 0.8179. Ridge Regression Model

```python
# Standardize Feature
from sklearn.preprocessing import StandardScaler
scaler = StandardScaler()
x_train_scaled = scaler.fit_transform(x_train)
x_test_scaled = scaler.transform(x_test)
```

```python
print('Training Data Shape:', x_train_scaled.shape)
print('Testing Data Shape:', x_test_scaled.shape)
```

```python
# Create Ridge Model
from sklearn.linear_model import Ridge

ridge_model = Ridge(alpha=1.0)

ridge_model.fit(x_train_scaled, y_train)
```

```python
y_train_ridge_pred = ridge_model.predict(x_train_scaled)
y_test_ridge_pred = ridge_model.predict(x_test_scaled)
```

```python
# Evaluate Ridge Model
from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score
import numpy as np

ridge_train_mae = mean_absolute_error(y_train, y_train_ridge_pred)
ridge_train_rmse = np.sqrt(mean_squared_error(y_train, y_train_ridge_pred))
ridge_train_r2 = r2_score(y_train, y_train_ridge_pred)

ridge_test_mae = mean_absolute_error(y_test, y_test_ridge_pred)
ridge_test_rmse = np.sqrt(mean_squared_error(y_test, y_test_ridge_pred))
ridge_test_r2 = r2_score(y_test, y_test_ridge_pred)

print("Ridge Training MAE:", ridge_train_mae)
print("Ridge Training RMSE:", ridge_train_rmse)
print("Ridge Training R²:", ridge_train_r2)

print("\nRidge Testing MAE:", ridge_test_mae)
print("Ridge Testing RMSE:", ridge_test_rmse)
print("Ridge Testing R²:", ridge_test_r2)
```

```python
# Testing different alpha values
from sklearn.linear_model import Ridge
from sklearn.metrics import mean_squared_error, r2_score
import numpy as np

alphas = [0.01, 0.1, 1, 10, 100]

ridge_results = []

for alpha in alphas:
    ridge = Ridge(alpha=alpha)
    ridge.fit(x_train_scaled, y_train)

    y_test_pred = ridge.predict(x_test_scaled)

    rmse = np.sqrt(mean_squared_error(y_test, y_test_pred))
    r2 = r2_score(y_test, y_test_pred)

    ridge_results.append([alpha, rmse, r2])

ridge_results
```

```python
from sklearn.linear_model import RidgeCV

alphas = [0.01, 0.1, 1, 10, 100]

ridge_cv = RidgeCV(
    alphas=alphas,
    cv=5,
    scoring='neg_mean_squared_error'
)
ridge_cv.fit(x_train_scaled, y_train)
```

```python
print('Best Alpha:', ridge_cv.alpha_)
```

```python
y_test_ridge_cv_pred = ridge_cv.predict(x_test_scaled)
```

```python
from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score
import numpy as np

ridge_cv_mae = mean_absolute_error(y_test, y_test_ridge_cv_pred)
ridge_cv_rmse = np.sqrt(mean_squared_error(y_test, y_test_ridge_cv_pred))
ridge_cv_r2 = r2_score(y_test, y_test_ridge_cv_pred)

print("Ridge CV Test MAE:", ridge_cv_mae)
print("Ridge CV Test RMSE:", ridge_cv_rmse)
print("Ridge CV Test R²:", ridge_cv_r2)
```

Output:

Ridge Regression was implemented to address potential multicollinearity and control coefficient magnitude through L2 regularization.

Five-fold cross-validation was used to select the optimal regularization parameter, and α = 10 was selected. On the test set, the Ridge model achieved an MAE of 0.0429, RMSE of 0.0610, and R² of 0.8183.

Compared with the OLS model, Ridge showed very similar performance but did not provide a meaningful improvement.

Therefore, OLS remains preferable for this dataset based on predictive performance, while Ridge provides the benefit of coefficient shrinkage and greater robustness to multicollinearity
